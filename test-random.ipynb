```python

import json
import runpy
import time

from discoveryworld.DiscoveryWorldAPI import DiscoveryWorldAPI

random_agent = runpy.run_path(
    "environment/discoveryworld/agents/RandomAgent.py"
)["randomAgent"]

api = DiscoveryWorldAPI(threadID=time.time_ns())
if not api.loadScenario("Combinatorial Chemistry", "Easy", randomSeed=0, numUserAgents=1):
    raise RuntimeError("Could not load Combinatorial Chemistry (Easy)")

random_agent(api, numSteps=26, seed=0)
print(json.dumps(api.getTaskScorecard(), indent=2))
```

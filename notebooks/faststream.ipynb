```python
from faststream.rabbit import RabbitBroker, TestRabbitBroker

broker = RabbitBroker()

@broker.subscriber("ben.inbox")
async def handle_message(body: str):
    print(f"Ben received: {body}")

async def demo():
    async with TestRabbitBroker(broker) as test_broker:
        await test_broker.publish("Ada found the sample", queue="ben.inbox")
```

```python
from faststream.rabbit import RabbitBroker, TestRabbitBroker

broker = RabbitBroker()

class TestPlayer:
    def __init__(self, name: str):
        self.name = name
        self.inbox = f"agent.{name.lower()}"
        self.received = []

    def handle_message(self, message: str):
        self.received.append(message)
        print(f"{self.name} received: {message}")


ada = TestPlayer("Ada")
ben = TestPlayer("Ben")

broker.subscriber(ada.inbox)(ada.handle_message)
broker.subscriber(ben.inbox)(ben.handle_message)
```

```python
async with TestRabbitBroker(broker) as test_broker:
    await test_broker.publish("task 2", queue=ada.inbox)
```

```python
from typesafe_sdk import TypeSafeClient, Choice, Noul, Score
import asyncio
from collections import defaultdict
from dataclasses import dataclass
import os
from dotenv import load_dotenv

client = TypeSafeClient(
    api_key=os.environ["OPENROUTER_API_KEY"],
    base_url="https://openrouter.ai/api",
)
```

```python
ticket = """
I was charged twice for my monthly subscription.
I can see two separate transactions on my bank statement.
"""

result = client.system_one(
    state=ticket,
    model="jev-latest",
    questions={
        "department": Choice(
            instructions="Which department should handle this ticket?",
            criteria={
                "billing": "Payments, invoices and duplicate charges",
                "technical": "Software faults and technical problems",
                "account": "Login, registration and account access",
                "other": "Issues that do not fit the other categories"
            }
        )
    }
)

answer = result.answers["department"]
print("Department:", answer.choice)
print("Confidence:", answer.confidence)
print("Confidence type:", type(answer.confidence))
```

```python
agents = {
    "agentX": "AgentX (agent_idx = 0):  Agentic player, its task is to complete a task on DiscoveryWorld environment" ,
    "agentY": "AgentY (agent_idx = 2):  Agentic player, its task is to complete a task on DiscoveryWorld environment" ,
    "agentZ": "AgentZ (agent_idx = 3):  Agentic player, its task is to complete a task on DiscoveryWorld environment" ,
    "broadcast": "broadcast:  when there is no one specific destination, messages should be routed/sent to bradcast" ,
    }
message_broadcast = """AgentX: hey guys, I am gonna submit my action in Crystal 2, need to verify if the resonance frequency follows a linear relation with the quantum gap. 
Can anyone help me with the other crystals and check this out?
"""
message_for_agentY = """agentX: I'm going to test crystal 2. @agentY, could you verify crystal 3 please?"""
message_for_agentX = "agentZ: hey agentX, I am not able to find crystal 1 and 2, any idea?"
message_for_agentZ = "can you explore south area agent_idx 3? I will go for east."
message_broadcast_2 = """Task description: 
You are at the Quantum Reactor Lab on Planet X. 
Quantum Crystals offer the potential to generate a great deal of power, 
but require their respective crystal reactors to be tuned to a specific frequency that appears unique for each crystal. 
Through great effort, a previous research scientist manually stumbled upon the correct frequencies for Crystal 1 and Crystal 2, which are in their respective reactors. 
Your task is to use science to figure out the correct frequency for Crystal 3, set it's reactor to the appropriate frequency, and install the crystal. 
Once all three crystals are installed, the reactor will be able to generate a great deal of power. 
To support your task, a scientific instrument is available in the lab. 
"""
```

```python
client = TypeSafeClient(
    api_key=os.environ["OPENROUTER_API_KEY"],
    base_url="https://openrouter.ai/api",
)

def route_message(message):
    result = client.system_one(
        state=message,
        model="jev-latest",
        questions={
            "destination": Choice(
                instructions="Which agent should handle this message?",
                criteria=agents
            )
        }
    )
    # print(result.answers["destination"].choice)
    print("destination:",result.answers["destination"].choice)
    print("Confidence:", result.answers["destination"].confidence)
    return result

route_message(message_broadcast)
route_message(message_for_agentY)
route_message(message_for_agentX)
route_message(message_for_agentZ)
route_message(message_broadcast_2)
```

```python
import asyncio

async def waiter(event, name):
    print(f'{name}: waiting for it ...')
    #agents should have a tool and wait
    await event.wait()
    print(f'{name}: ... got it!')

async def main():
    event = asyncio.Event()
    waiter_task = asyncio.create_task(waiter(event, "first"))

    # Signal after 5s, while the wait below only lasts 3s.
    async def set_later():
        await asyncio.sleep(5)
        event.set()

    setter = asyncio.create_task(set_later())

    try:
        return await asyncio.wait_for(waiter_task, timeout=3)
    except TimeoutError:
        setter.cancel()
        return "No message arrived. You can continue investigating or wait again."

await main()
```

```python
import asyncio


async def waiter(event, name):
    print(f'{name}: waiting for it ...')
    await event.wait()
    print(f'{name}: ... got it!')

async def main():
    # Create an Event object.
    event = asyncio.Event()
    event2 = asyncio.Event()
    # Spawn tasks with different parameters.
    waiter_task = asyncio.create_task(waiter(event, "first"))
    waiter_task2 = asyncio.create_task(waiter(event2, "second"))
    tasks = [waiter_task, waiter_task2]
    
    # Sleep for 1 second and set the event.
    await asyncio.sleep(3)
    
    
    event.set()
    event2.set()
    
    # Wait until both tasks are finished.
    await asyncio.gather(*tasks)

# asyncio.run(main())
await main()
```

```python
from tempfile import TemporaryDirectory

from discoveryworld.Agent import CrystalReactor
from discoveryworld.DiscoveryWorldAPI import DiscoveryWorldAPI

with TemporaryDirectory() as frames:
    api = DiscoveryWorldAPI()
    api.FRAME_DIR = frames
    assert api.loadScenario("Reactor Lab", "Easy", randomSeed=0)

    reactor = next(
        obj for obj in api.world.getAllWorldObjects()
        if isinstance(obj, CrystalReactor)
    )

    result = api.performAgentAction(0, {
        "action": "TELEPORT_TO_OBJECT",
        "arg1": reactor.uuid,
    })
    assert result["success"], result
    api.tick()

    result = api.performAgentAction(0, {
        "action": "USE",
        "arg1": reactor.uuid,
        "arg2": reactor.parentContainer.uuid,  # The table under the reactor
    })
    assert result["success"], result
    api.tick()

    print("Live dialogue state:", api.isAgentInDialog(0))
    print("Observed dialogue box:", api.getAgentObservation(0)["ui"]["dialog_box"])

    result = api.performAgentAction(0, {"chosen_dialog_option_int": 1})
    print("Option 1 accepted:", result["success"])
```

```python
from tempfile import TemporaryDirectory

from discoveryworld.Agent import CrystalReactor
from discoveryworld.DiscoveryWorldAPI import DiscoveryWorldAPI


def act(api, **payload):
    result = api.performAgentAction(0, payload)
    assert result["success"], result
    api.tick()


def show(api, label):
    box = api.getAgentObservation(0)["ui"]["dialog_box"]
    print(
        f"{label}: live={api.isAgentInDialog(0)}, "
        f"observed={box['is_in_dialog']}, "
        f"options={len(box.get('dialogOptions', {}))}"
    )


for opener in ("TALK", "USE"):
    with TemporaryDirectory() as frames:
        api = DiscoveryWorldAPI()
        api.FRAME_DIR = frames
        assert api.loadScenario("Reactor Lab", "Easy", randomSeed=0)

        reactor = next(
            obj for obj in api.world.getAllWorldObjects()
            if isinstance(obj, CrystalReactor)
        )
        act(api, action="TELEPORT_TO_OBJECT", arg1=reactor.uuid)

        if opener == "TALK":
            act(api, action="TALK", arg1=reactor.uuid)
        else:
            act(api, action="USE", arg1=reactor.uuid,
                arg2=reactor.parentContainer.uuid)

        show(api, f"{opener} / opened")

        act(api, chosen_dialog_option_int=1)  # Adjust frequency and stay in the menu
        show(api, f"{opener} / after option 1")

        act(api, chosen_dialog_option_int=9)  # Exit
        show(api, f"{opener} / after Exit")

        act(api, action="TELEPORT_TO_OBJECT", arg1=reactor.uuid)
        show(api, f"{opener} / after ordinary action")
```
